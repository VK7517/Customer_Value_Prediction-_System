```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np 
```

```python
data = pd.read_csv("../data/raw/Online Retail.csv")
```

```python
data.head()
```

```python
data = data.dropna(subset=["CustomerID"])
```

```python
data.shape
```

```python
data["InvoiceDate"] = data["InvoiceDate"].str.replace(".",":")
```

```python
data["InvoiceDate"] = pd.to_datetime(data["InvoiceDate"],format="%d-%m-%Y %H:%M")
```

```python
data["CustomerID"] = data["CustomerID"].astype("int")
```

```python
data["Iscancelled"] = data["InvoiceNo"].str.startswith("C")
```

```python
data["Iscancelled"].value_counts()
```

```python
data = data[~data["Iscancelled"]].copy()
```

```python
data = data[data["Quantity"]>0].copy()
```

```python
data = data[data["UnitPrice"]>0].copy()
```

```python
data["TransactionRevenue"] = data["Quantity"] * data["UnitPrice"] 
```

```python
data["TransactionRevenue"]
```

```python

data.isnull().sum()
```

```python
data.to_csv("../data/processed/Cleaned_Online_Retail.csv",index=False)
```
