```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np 
```

```python
data = pd.read_csv("../data/raw/Online Retail.csv")
```

```python
data.head()
```

```python
data["InvoiceDate"] = data["InvoiceDate"].str.replace(".",":")
```

```python
data["InvoiceDate"] = pd.to_datetime(data["InvoiceDate"],format="%d-%m-%Y %H:%M")
```

```python
data["InvoiceDate"].value_counts()
```

```python
data["Month"] = data["InvoiceDate"].dt.month
```

```python
data["Year"] = data["InvoiceDate"].dt.year
```

```python
data["Day"] = data["InvoiceDate"].dt.day
```

```python
df = data.groupby(["CustomerID","Month"])["Day"].agg(
    No_of_days = "count" 
)
# How many days customer do purchase 
No_of_days_customer_purchase = df.groupby("CustomerID")["No_of_days"].sum()
```
